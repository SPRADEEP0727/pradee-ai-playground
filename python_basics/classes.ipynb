# Python Classes - A Beginner's Guide

Welcome! In this notebook, we will learn about **Classes** in Python.

## What is a Class?

A **class** is like a **blueprint** for creating things (called *objects*).

Think of it like this:
- A blueprint of a house is the **class**.
- The actual houses built from that blueprint are the **objects**.

Real-life examples:
- `Phone` is a class. *My iPhone*, *Your Samsung* are objects (phones).
- `Car` is a class. *My Honda*, *Your Toyota* are objects (cars).
- `Student` is a class. *Ravi*, *Priya*, *John* are objects (students).

## 1. Creating Your First Class

We use the keyword `class` to create a class.

**Syntax:**
```python
class ClassName:
    # body of the class
```

````python
# A very simple class
class Phone:
    brand = "Samsung"

# Create an object (instance) of the class
my_phone = Phone()

# Access the data inside the object
print(my_phone.brand)
````

## 2. The `__init__` Method (Constructor)

The `__init__` method runs **automatically** when you create a new object.
It is used to set the starting values of the object.

- `self` refers to the object itself.
- Anything starting with `self.` belongs to the object.

````python
class Student:
    def __init__(self, name, age):
        self.name = name   # store the name
        self.age = age     # store the age

# Create student objects
s1 = Student("Ravi", 15)
s2 = Student("Priya", 16)

print(s1.name, "is", s1.age, "years old")
print(s2.name, "is", s2.age, "years old")
````

## 3. Adding Methods (Functions inside a Class)

A **method** is a function that belongs to a class.
It tells what the object can **do**.

````python
class Lamp:
    def __init__(self, color):
        self.color = color
        self.is_on = False

    def turn_on(self):
        self.is_on = True
        print("The", self.color, "lamp is now ON.")

    def turn_off(self):
        self.is_on = False
        print("The", self.color, "lamp is now OFF.")

# Create a lamp
my_lamp = Lamp("yellow")

# Call the methods
my_lamp.turn_on()
my_lamp.turn_off()
````

## 4. A Slightly Bigger Example - A Car

Let's build a `Car` class that has properties (data) and behaviors (methods).

````python
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color
        self.speed = 0

    def start(self):
        print(self.brand, "is starting...")

    def accelerate(self, increase):
        self.speed = self.speed + increase
        print(self.brand, "speed is now", self.speed, "km/h")

    def stop(self):
        self.speed = 0
        print(self.brand, "has stopped.")

# Create a car object
my_car = Car("Honda", "Red")

my_car.start()
my_car.accelerate(40)
my_car.accelerate(20)
my_car.stop()
````

## 5. Many Objects from One Class

From a single class, you can create **as many objects as you want**.
Each object has its own data.

````python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def info(self):
        print("Book:", self.title, "| Author:", self.author)

b1 = Book("Harry Potter", "J.K. Rowling")
b2 = Book("The Jungle Book", "Rudyard Kipling")
b3 = Book("Wings of Fire", "A.P.J. Abdul Kalam")

b1.info()
b2.info()
b3.info()
````

## 6. Inheritance - Reusing a Class

**Inheritance** lets a new class take all the features of an existing class.

- The original class is called the **parent** (or base) class.
- The new class is called the **child** (or derived) class.

Example: A `Bike` is a `Vehicle`, so `Bike` can inherit from `Vehicle`.

````python
# Parent class
class Vehicle:
    def __init__(self, name):
        self.name = name

    def move(self):
        print(self.name, "is moving.")

# Child class - inherits from Vehicle
class Bike(Vehicle):
    def ring_bell(self):
        print(self.name, "rings: Tring Tring!")

b = Bike("Hero Bike")
b.move()        # comes from Vehicle
b.ring_bell()   # comes from Bike
````

## 7. Quick Recap

| Term | Meaning |
|------|---------|
| `class` | Blueprint for creating objects |
| `object` | A real thing built from the class |
| `__init__` | Special method that runs when an object is created |
| `self` | Refers to the current object |
| `method` | A function inside a class |
| `inheritance` | One class taking features from another |
